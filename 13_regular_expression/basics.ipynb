## functions in re
1. re.findall()<br>
Returns all non-overlapping matches of a pattern in the string as a list. It scans the string from left to right.<br><br>
2. re.compile()<br>
Compiles a regex into a pattern object, which can be reused for matching or substitutions.<br><br>
3. re.split() <br>
Splits a string wherever the pattern matches. The remaining characters are returned as list elements.<br>
re.split(pattern, string, maxsplit=0, flags=0)<br>
maxsplit (optional): Limits the number of splits. Default is 0 (no limit).<br>
flags (optional): Apply regex flags like re.IGNORECASE.<br><br>
4. re.sub()<br>
The re.sub() function replaces all occurrences of a pattern in a string with a replacement string.<br>
 re.sub(pattern, repl, string, count=0, flags=0)<br>
pattern: The regex pattern to search for.<br>
repl: The string to replace matches with.<br>
string: The input string to process.<br>
count (optional): Maximum number of substitutions (default is 0, which means replace all).<br>
flags (optional): Regex flags like re.IGNORECASE.<br><br>
5. re.subn()<br>
re.subn() function works just like re.sub(), but instead of returning only the modified string, it returns a tuple: (new_string,number_of_substitutions)<br><br>
6. re.escape()<br>
re.escape() function adds a backslash (\) before all special characters in a string. This is useful when you want to match a string literally, including any characters that have special meaning in regex (like ., *, [, ], etc.).<br><br>
7. re.search()<br>
The re.search() function searches for the first occurrence of a pattern in a string. It returns a match object if found, otherwise None.<br><br>

```python
import re
string = """Hello my Number is 123456789 and
            my friend's number is 987654321"""

# findall - for finding all sequences of one or more digits in the given string.
pattern='\d+'
print(re.findall(pattern,string))

# compile -  for searching pattern [a-e] matches all lowercase letters between 'a' and 'e'
pattern=re.compile('[a-e]')
pattern.findall(string)

# split
print(re.split('\W+', 'Words, words , Words'))
print(re.split('\W+', "Word's words Words"))
print(re.split('\W+', 'On 12th Jan 2016, at 11:02 AM'))
print(re.split('\d+', 'On 12th Jan 2016, at 11:02 AM'))
print(re.split('\d+', 'On 12th Jan 2016, at 11:02 AM', 1))
print(re.split('[a-f]+', 'Aey, Boy oh boy, come here', flags=re.IGNORECASE))
print(re.split('[a-f]+', 'Aey, Boy oh boy, come here'))

# sub
print(re.sub('ub', '~*', 'Subject has Uber booked already', flags=re.IGNORECASE))

# Case-sensitive replacement of all 'ub'
print(re.sub('ub', '~*', 'Subject has Uber booked already'))

# Replace only the first 'ub', case-insensitive
print(re.sub('ub', '~*', 'Subject has Uber booked already', count=1, flags=re.IGNORECASE))

# Replace "AND" with "&", ignoring case
print(re.sub(r'\sAND\s', ' & ', 'Baked Beans And Spam', flags=re.IGNORECASE))

#subn
print(re.subn('ub', '~*', 'Subject has Uber booked already', count=1, flags=re.IGNORECASE))

# escape
print(re.escape("This is Awesome even 1 AM"))
print(re.escape("I Asked what is this [a-9], he said \t ^WoW"))

# search
regex = r"([a-zA-Z]+) (\d+)"
match = re.search(regex, "I was born on June 24")

if match:
    print("Match at index %s, %s" % (match.start(), match.end()))
    print("Full match:", match.group(0))
    print("Month:", match.group(1))
    print("Day:", match.group(2))
else:
    print("The regex pattern does not match.")
```

## Metacharecters in regex

| MetaCharacter | Description |
|---------------|-------------|
| `\`           | Used to drop the special meaning of the character following it |
| `[]`          | Represents a character class |
| `^`           | Matches the beginning |
| `$`           | Matches the end |
| `.`           | Matches any character except newline |
| `|`           | OR operator (matches with any of the characters separated by it) |
| `?`           | Matches zero or one occurrence |
| `*`           | Any number of occurrences (including 0 occurrences) |
| `+`           | One or more occurrences |
| `{}`          | Indicates the number of occurrences of the preceding regex to match |
| `()`          | Encloses a group of regex |


## Charecter matching
| Set / Pattern  | Description |
|----------------|-------------|
| `\{n,\}`       | Quantifies the preceding character or group and matches at least **n** occurrences |
| `*`            | Quantifies the preceding character or group and matches zero or more occurrences |
| `[0123]`       | Matches the specified digits (0, 1, 2, or 3) |
| `[^arn]`       | Matches any character **except** `a`, `r`, and `n` |
| `\d`           | Matches any digit (0–9) |
| `[0-5][0-9]`   | Matches any two-digit number from `00` to `59` |
| `\w`           | Matches any alphanumeric character (`a–z`, `A–Z`, `0–9`, or `_`) |
| `[a-n]`        | Matches any lowercase alphabet between `a` and `n` |
| `\D`           | Matches any **non-digit** character |
| `[arn]`        | Matches if one of the specified characters (`a`, `r`, or `n`) is present |
| `[a-zA-Z]`     | Matches any alphabet between `a–z` or `A–Z` (lowercase or uppercase) |
| `[0-9]`        | Matches any digit between `0` and `9` |


## Special sequence
| Pattern | Description |
|---------|-------------|
| `\A`    | Matches if the string begins with the given character |
| `\b`    | Matches if the word begins or ends with the given character (`\bword` = beginning, `word\b` = ending) |
| `\B`    | Opposite of `\b`, i.e., the string should **not** start or end with the given regex |
| `\d`    | Matches any decimal digit, equivalent to `[0-9]` |
| `\D`    | Matches any non-digit character, equivalent to `[^0-9]` |
| `\s`    | Matches any whitespace character |
| `\S`    | Matches any non-whitespace character |
| `\w`    | Matches any alphanumeric character, equivalent to `[a-zA-Z0-9_]` |
| `\W`    | Matches any non-alphanumeric character |
| `\Z`    | Matches if the string ends with the given regex |

```python
Paragraph=r"""On 12-08-2025, John’s email was john_doe99@example.com
, and his backup email was j.doe@work.org
.
He created a password: Pass@1234.
Later, he saved phone numbers: +91-9876543210 and (080)1234567.
The folder path was C:\Users\John\Documents\files and the website he visited was https://www.example-site.com/page?id=42
.
At the end of the day, the log file contained entries like:
[INFO] Process started at 10:30 AM
[ERROR] Failed at 10:45 AM
[DEBUG] Retrying..."""
```

```python
# Digits → Write a regex to extract all the dates (like 12-08-2025). (\d)

re.findall("\d{2}-\d{2}-\d{4}",Paragraph)
```

```python
#Word characters → How would you match only usernames from emails (john_doe99, j.doe)? (\w, .)
re.findall("[\w]+(?=@)",Paragraph) # matches full regex pattern but catches only part inside ()
```

```python
# Character sets → Write a regex to find all email domains (example.com, work.org). ([])

re.findall("@([\w]+.\w+)",Paragraph)
```

```python
# Quantifiers → How would you match all phone numbers where digits repeat a certain number of times? (+, {})

re.findall("\d{10}",Paragraph)
```

```python
# scaping special chars → Write a regex to detect Windows file paths like C:\Users\.... (\\)
re.findall("[A-Z]:\\[A-Za-z0-9\\]+",Paragraph)
```

```python
# Anchors → How do you match only lines that start with [ERROR]
re.findall("^\[ERROR\].*",Paragraph,re.MULTILINE)
```

```python
# End anchor → How to find lines ending with AM?
re.findall(".*AM",Paragraph,re.MULTILINE)
```

```python
# Alternation → Write a regex to match either [INFO] or [DEBUG]
re.findall("\[INFO\]|\[DEBUG\]",Paragraph)
```

```python
# Groups → How to capture the protocol (https) separately from the domain in the URL?
re.findall("(\w+)://([\w.-]+)",Paragraph)
```

```python
# Greedy vs Lazy → How do you match only https://www.example-site.com without including ?id=42?
re.findall("https://[\w.-]+\.com",Paragraph)
```

```python
# Whitespace → Find all timestamps with spaces inside (10:30 AM)
re.findall("\d{2}:\d{2}\sAM",Paragraph)
```

```python
# Lookahead → Match all numbers that are followed by AM
re.findall("(\d+)\sAM",Paragraph)
```

```python
# Extract only times (like 10:30) that come after [INFO] or [ERROR]
re.findall("\[INFO\].*(\d{2}:\d{2})|\[ERROR\].*(\d{2}:\d{2})",Paragraph)
```

# special symbols

1. ? (Question mark)<br>
It has two main uses: <br><br>
&nbsp;1.1) Optional → As an Optional Quantifier<br>
&nbsp;It means “the preceding character/group may appear 0 or 1 time.”<br>
&nbsp;eg - colou?r will match both color and colour<br><br>
1.2) As a Lazy Modifier<br>
👉 It makes *, +, {} match as little as possible instead of as much as possible.<br>
eg- string = https://a.com/abc.com/xyz.com<br>
    pattern = https://.*\.com , output = https://a.com<br>
    pattern = https://.*?\.com , output = https://a.com/abc.com/xyz.com<br><br>
1.3) In Special Sequences (with : or =, <)<br><br>
When ? is inside parentheses, it changes meaning:<br><br>
(?:...) → non-capturing group (just groups things, doesn’t save)<br>
(?=...) → positive lookahead (must be followed by …)<br>
(?!...) → negative lookahead (must not be followed by …)<br>
(?<=...) → positive lookbehind (must be preceded by …)<br>
(?<!...) → negative lookbehind (must not be preceded by …)<br><br>
2. () (Parentheses)<br><br>
2.1) Capturing groups → Extract parts of a match.<br><br>
2.2) Non-capturing groups (?:...) → Just grouping, don’t capture.<br>
eg (?:INFO|DEBUG) -> Matches either, but doesn’t store the group.<br><br>
2.3) Alternation with groups
eg: (INFO|DEBUG|ERROR) -> Matches one of them.
