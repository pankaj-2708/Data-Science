```python
import time
start_time=time.time()

def heavy_function(id):
    print("sleeping for 1 sec , id = ",id)
    time.sleep(1)
    print("woken up id = ",id)

for i in range(10):
    heavy_function(i)

print(f"time_taken={time.time()-start_time}")
```

```python
import time
import threading

start_time=time.time()

def heavy_function(id):
    print("sleeping for 1 sec , id = ",id)
    time.sleep(1)
    print("woken up id = ",id)

# creating 10 threads
all_threads=[threading.Thread(target=heavy_function,args=[i]) for i in range(10)]
for i in range(10):
    all_threads[i].start()

# joining all threads -> here it waits for all threads to complete and then move forward
for i in range(10):
    all_threads[i].join()

print(f"time_taken={time.time()-start_time}")
```

```python
# Python threads share the same memory space, and all threds modify the same variable balance.
# Even though Python has a Global Interpreter Lock (GIL), it does not guarantee atomicity for operations.
# for this loc is used


import time
import threading

start_time=time.time()

lock=threading.Lock()
balance=200

def deposit(n):
    global balance
    for i in range(n):
        with lock:
            balance+=1

def withdraw(n):
    global balance
    for i in range(n):
        with lock:
            balance-=1

thread1=threading.Thread(target=deposit,args=[10000])
thread2=threading.Thread(target=withdraw,args=[10000])

thread1.start()
thread2.start()

thread1.join()
thread2.join()

print(balance)
```

```python
from concurrent.futures import ThreadPoolExecutor

start_time=time.time()


# Create a thread pool with 2 workers
with ThreadPoolExecutor(max_workers=10) as executor:
    # Submit 10 tasks to run in parallel
    for i in range(10):
        executor.submit(heavy_function,i)
print(f"time_taken={time.time()-start_time}") 
```

```python
from concurrent.futures import ThreadPoolExecutor

start_time=time.time()
# locks syntax are same in both of them and are imported from threading only
# Create a thread pool with 2 workers
with ThreadPoolExecutor(max_workers=2) as executor:
    # Submit two tasks to run in parallel
    executor.submit(deposit,10000)
    executor.submit(withdraw,10000)
print(balance)
```

```python
# Solution -> using shared memory between processees 
# shared memory can be of three types -> 1) array 2) value 3) queue
#  read this article - https://www.geeksforgeeks.org/python/multiprocessing-python-set-2/
```

```python
import time
import multiprocessing

    
def square(arr,result):
    for idx,n in enumerate(arr):
        result[idx]=n * n
    print ("result in square", result[:])
    
def cube(arr,result):
    for idx,n in enumerate(arr):
        result[idx+5]=n * n*n
    print ("result in cube", result[:])

# wont work withon if __name__ == "__main__"
if __name__ == "__main__":
    start_time = time.time()

    arr=[2, 3, 4, 5, 6]
    
    # i means integer and d means double 10 is the size of the array
    result=multiprocessing.Array('i', 10)  # Shared array among processes
    
    p1= multiprocessing.Process(target=square, args=[arr,result])
    p2= multiprocessing.Process(target=cube, args=[arr,result])
        
    p1.start()
    p2.start()
    p1.join()
    p2.join()
    print ("Final result", result[:])

# output 
# (ML) C:\Users\panka\DSCampusX\14_adv_python>python -u "c:\Users\panka\DSCampusX\14_adv_python\multiprocess.py"
# result in square [4, 9, 16, 25, 36, 0, 0, 0, 0, 0]
# result in cube [4, 9, 16, 25, 36, 8, 27, 64, 125, 216]
# Final result [4, 9, 16, 25, 36, 8, 27, 64, 125, 216]
```

```python
import time
import multiprocessing

    
def square(arr,result,v):
    for idx,n in enumerate(arr):
        result[idx]=n * n
        v.value=v.value+n*n
    print ("result in square", result[:])
    print ("value in square", v.value)
    
    
def cube(arr,result,v):
    for idx,n in enumerate(arr):
        result[idx+5]=n * n*n
        v.value=v.value+n*n*n
    print ("result in cube", result[:])
    print ("value in cube", v.value)

# wont work withon if __name__ == "__main__"
if __name__ == "__main__":
    start_time = time.time()

    arr=[2, 3, 4, 5, 6]
    
    # i means integer and d means double 10 is the size of the array
    result=multiprocessing.Array('i', 10)  # Shared array among processes
    
    v=multiprocessing.Value('i', 0)  # Shared value among processes
    p1= multiprocessing.Process(target=square, args=[arr,result,v])
    p2= multiprocessing.Process(target=cube, args=[arr,result,v])
        
    p1.start()
    p2.start()
    p1.join()
    p2.join()
    print ("Final result ", result[:])
    print ("Final value", v.value)

# (ML) C:\Users\panka\DSCampusX\14_adv_python>python -u "c:\Users\panka\DSCampusX\14_adv_python\multiprocess.py"
# result in square [4, 9, 16, 25, 36, 0, 0, 0, 0, 0]
# value in square 90
# result in cube [4, 9, 16, 25, 36, 8, 27, 64, 125, 216]
# value in cube 530
# Final result  [4, 9, 16, 25, 36, 8, 27, 64, 125, 216]
# Final value 530
```

# Using a Manager (High-Level Shared Objects)
<img src="https://media.geeksforgeeks.org/wp-content/uploads/multiprocessing-python-3.png">
**Server process : Whenever a python program starts, a server process is also started. From there on, whenever a new process is needed, the parent process connects to the server and requests it to fork a new process. A server process can hold Python objects and allows other processes to manipulate them using proxies. multiprocessing module provides a Manager class which controls a server process. Hence, managers provide a way to create data that can be shared between different processes.**

```python
from multiprocessing import Process, Manager

def worker(shared_list, shared_dict):
    shared_list.append(10)
    shared_dict["x"] = 42

if __name__ == "__main__":
    with Manager() as manager:
        shared_list = manager.list()
        shared_dict = manager.dict()

        p1 = Process(target=worker, args=(shared_list, shared_dict))
        p2 = Process(target=worker, args=(shared_list, shared_dict))

        p1.start(); p2.start()
        p1.join(); p2.join()

        print(shared_list)  # [10, 10]
        print(shared_dict)  # {'x': 42}
```
