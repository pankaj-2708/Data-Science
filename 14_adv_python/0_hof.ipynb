```python
# lambda fn are used with higher order function
# higher order functions are functions that take a fn as input

def higher_order(func,l):
    result=0
    for i in l:
        if func(i):
            result+=i
    return result

l=[1,2,3,4,5,6]
even_sum=lambda x:x%2==0
odd_sum=lambda x:x%2==1

print("even sum " , higher_order(even_sum ,l))
print("odd sum " , higher_order(odd_sum ,l))
```

```python
# python generally used higher order fn 

# map 
# map require an fn and an list/dict etc
# it apply that fn to each member of above iterable
# it return an map object which can be converted to list using list

print(list(map(lambda x: x%2==0,l)))

# filter
# filter also require an fn and iterable to work
# it filters out and return only those values which return true for the fn 

print(list(filter(lambda x:x>4,l)))


# reduce 
# it is inside functools
# it gives inputs in pairs to given function and keep store the output as the new item

from functools import reduce

# it will return the max element
print(reduce(lambda x,y:x if x>y else y,l))

# it will return the min element
print(reduce(lambda x,y:x if x<y else y,l))
```
