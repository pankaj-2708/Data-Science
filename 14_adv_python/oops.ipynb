```python
class Fraction:
    def __init__(self,num,den):
        self.num=num
        self.den=den

    # __str__ is a magic method which is automatically called when we put our object inside print function
    def __str__(self):
        return f"{self.num}/{self.den}"

    # __add__ is another magic method , __sub__ ,__mul__ also exsist
    def __add__(self,other):
        new_frac=Fraction(self.num * other.den + self.den * other.num, self.den * other.den)
        return new_frac
```

```python
a=Fraction(3,5)
print(a)
b=Fraction(4,5)
print(a+b)
```

```python
# inheritance

class Parent:
    def __init__(self,parent_name):
        self.parent_name=parent_name
    def get_parent_name(self):
        return self.parent_name

    def demo_func(self):
        print("parent's demo _func called")


# if child dont have its contructor then constructor of parent will be called
# else constructor of child is called
class child(Parent):
    def __init__(self,child_name,parent_name):
        # by super keyword we can call parents constructor
        super().__init__(parent_name)
        self.child_name=child_name
        
        
    def get_child_name(self):
        return self.child_name

    def demo_func(self):
        print("child's demo _func called")
        # by super keyword parent's functions are called
        super().demo_func()
        
child1=child("child","parent")
child1.get_child_name()
```

## I. Introduction to OOP and Fundamental Concepts

### A. The Need for Object-Oriented Programming (OOP)
1.  OOP is a distinct style of writing code.
2.  In modern development, real-life problems are often solved by converting real-world experiences into software (e.g., online banking, ticket booking).
3.  Traditional, non-OOP coding can lead to a huge, confusing code base that makes adding new features or making modifications extremely difficult (as experienced when building a LinkedIn-like platform, taking six to eight months).
4.  OOP helps simplify code management and makes applications more organized and optimized.

### B. Objects and Classes
1.  In Python, **everything is an object**.
2.  Every object is associated with a **Class**.
3.  **Data Types are Classes** (e.g., List, String, Integer).
4.  When you create a variable of a data type (e.g., `l =`), that variable (`l`) is the **Object**. Therefore, an object is an **instance** of a class.
5.  **Class:** A class acts as a **blueprint** that dictates how an object will behave. Defining a class requires significantly more coding effort than creating an object from it.
6.  **Custom Data Types:** The core fundamental feature of OOP is that a programmer can create their own data types (classes) tailored to specific application needs (e.g., a data type specifically designed for a LinkedIn platform or a `Fraction` data type).

### C. Structure of a Class
A class contains two important components:
1.  **Data (Properties/Attributes):** Descriptive information about the object (e.g., a car's color, brand, or engine type).
2.  **Functions (Behaviors/Methods):** The actions the object can perform (e.g., calculating mileage or speed).

### D. Naming Conventions & Representation
1.  **Class Names:** Should be written in **PascalCase** (e.g., `ClassName`).
2.  **Variables/Methods:** Should be written in **snake_case** (e.g., `variable_name` or `function_name`).
3.  **UML Class Diagrams:** Classes can be represented diagrammatically, showing the Class Name at the top, Attributes (Data) in the middle, and Methods (Functions) at the bottom.
    *   **Access Modifiers:** Plus (`+`) indicates a public member (accessible outside the class); minus (`-`) indicates a private member (not easily accessible outside the class).

### E. Methods vs. Functions
1.  **Function:** A block of code defined generally outside any class and is available globally (e.g., the `len()` function).
2.  **Method:** A function defined **inside a class** that can only be accessed by an object of that class (e.g., `list.append()`). Programmers should use the term "method" when referring to functions inside a class.

### F. Variables within a Class
1.  **Instance Variables:** These variables are defined **inside the constructor** (`__init__`) and their value is **different for every object**. (Example: the PIN or balance of different ATM customers).
2.  **Static Variables (Class Variables):** These variables are defined **outside the constructor** and their value is **shared among all objects** of the class. They are useful for tracking shared information, like a counter for how many objects have been created, or a bank's IFSC code.

### G. The `__init__` Method (Constructor)
1.  **Definition:** `__init__` is a **special method** (a **Magic Method** or **Dunder Method**) that acts as the constructor in Python. It must be named exactly `__init__` with double underscores.
2.  **Execution:** The code inside the constructor is **automatically executed** as soon as an object of that class is created.
3.  **Purpose:** The constructor should handle configuration tasks that must run immediately and are outside the user's control, such as establishing connectivity (internet, database, GPS).

### H. The `self` Keyword
1.  **Identity:** `self` is a parameter passed to every instance method, and it represents the **object that is currently calling the method**. The memory address (ID) of `self` is identical to the memory address of the object itself.
2.  **Mechanism:** When an object calls a method (e.g., `sbi.deposit()`), Python passes the object (`sbi`) by default to the method, and this object is received by the `self` parameter.
3.  **Necessity:** Within OOP, a method cannot directly access other methods or data members within its own class; it must do so through an object. Since `self` holds the reference to the current object, methods use `self.attribute` or `self.method()` to interact internally.

## II. The Pillars of OOP

### A. Encapsulation
1.  **Need for Protection:** Allowing direct, public access to data members (e.g., `sbi.balance = "string"`) is poor practice because external manipulation can corrupt the data type and break existing operational logic (like deposit/withdrawal functions).
2.  **Hiding Data (Privacy):** In Python, data members are hidden by prepending a **double underscore (`__`)** to the attribute name (e.g., `__balance`).
3.  **Getters and Setters:** To allow controlled access to hidden data, public methods should be provided:
    *   **Getter:** A method to retrieve the value of the attribute.
    *   **Setter:** A method to change the value of the attribute.
4.  **Data Protection:** Setters are essential because they allow the programmer to include **validation logic** (e.g., checking if the new PIN is a string or an integer) before the attribute is updated, thereby protecting the integrity of the data.
5.  **Definition:** Encapsulation is the practice of combining the data member and its associated access methods (Getters and Setters) together, allowing the data member to be used while maintaining its protection.

### B. Inheritance (Is-A Relationship)
1.  **Definition:** Inheritance allows a Child class (Subclass) to derive or access properties and methods from a Parent class (Superclass), creating an **Is-A relationship** (e.g., a `Student` *is a* `User`).
2.  **Key Benefit:** **Code Reusability**. Common features (like Login/Registration) can be placed in a Parent class and reused by all Child classes, preventing repetition.
3.  **Direction:** Inheritance is uni-directional: Child classes inherit from Parents, but Parents **cannot** inherit features from their Children.
4.  **Inherited Components:** Public data members, methods, and the constructor are inherited.
5.  **Non-Inherited Components:** Private members (those prefixed with `__`) are **not** inherited by Child classes.
6.  **Constructor Rules:**
    *   If a Child class lacks a constructor, the Parent's constructor is automatically invoked upon object creation.
    *   If a Child class defines its own constructor, the Parent's constructor is **not** called automatically.
7.  **Types of Inheritance (Python supports all):**
    *   **Single Level:** One parent, one child.
    *   **Multi-Level:** A chain of inheritance (e.g., Class C inherits from B, which inherits from A).
    *   **Hierarchical:** One parent has multiple children (e.g., User is Parent to Student and Instructor).
    *   **Multiple Inheritance:** A child inherits from two or more separate parent classes (Python supports this, unlike Java).
        *   **Method Resolution Order (MRO):** In Multiple Inheritance, if the parents have conflicting methods, the method executed is determined by the order in which the parent classes were listed in the child's class definition.

### C. The `super()` Keyword
1.  `super()` is used inside a Child class to explicitly access and invoke methods or the constructor of its immediate Parent class.
2.  It is critical when a Child class has its own constructor but needs to delegate the initialization of common attributes (like brand or price) to the Parent's constructor, then handle its own specific attributes (like RAM or OS).
3.  `super()` cannot be used outside of a class method.

### D. Polymorphism (Multiple Forms)
Polymorphism refers to the ability of code to take on multiple forms.
1.  **Method Overriding:** Occurs when a Child class redefines a method that already exists in its Parent class. When the method is called, the Child's version takes precedence. This ensures that the object executes its "own" function over the inherited one.
2.  **Operator Overloading:** Customizing the default mathematical or logical behavior of standard Python operators (`+`, `-`, `*`) for custom objects. This is achieved using corresponding **Magic Methods** (e.g., `__add__` for `+`, `__mul__` for `*`, etc.).
3.  **Method Overloading:** Defining multiple methods with the same name but different input parameters. While technically not supported natively by Python (defining the same method name overwrites previous definitions), the effect can be achieved by using **default arguments** and internal conditional logic to handle different numbers of inputs.

### E. Other Class Relationships
1.  **Aggregation (Has-A Relationship):** A relationship where one class contains an object of another class as an attribute. (Example: A `Customer` class might have an `Address` object stored within it, indicating the Customer **Has an** Address).

## III. Practical Implementation Concepts

### A. Magic Methods and Triggers
Magic methods (Dunder Methods) are special functions that are automatically executed by Python under specific circumstances, rather than being called explicitly by the object.
*   `__str__`: Called when an object is passed to the `print()` function. Used to define how the object should be displayed as a string (e.g., 5/6 for a fraction).
*   `__add__`, `__sub__`, `__mul__`, `__truediv__`: Called when the operators `+`, `-`, `*`, or `/` are used between two objects, respectively.

### B. Passing Objects as Arguments and Returns
Class objects can be passed as arguments to functions, just like any built-in data type (like integers or lists). They can also be returned from a function.

### C. Mutability of Custom Objects
Objects created from custom classes are typically **mutable** (like Python lists and dictionaries). If a custom object is passed to a function, and that function modifies the object's attributes, the changes will be permanent in the original object outside the function.
