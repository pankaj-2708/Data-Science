## Lecture Notes: Iterators, Iterables, and Iteration in Python

### 1. Core Definitions

| Term | Definition | Role |
| :--- | :--- | :--- |
| **Iteration** | The general process of taking and processing each item of a sequence one after another. This occurs anytime a loop is used (explicitly or implicitly). | The *process* of moving through data. |
| **Iterable** | An object over which one can perform iteration (i.e., run a loop). Built-in examples include lists, tuples, dictionaries, sets, and range objects. | The *object* capable of being iterated over. |
| **Iterator** | An object that allows traversal through a sequence of data. It is the mechanism that enables iteration. | The *tool* that performs the iteration. |

### 2. Key Characteristics and Memory Efficiency

The defining characteristic of an Iterator is that it **does not need to load the entire sequence of data into memory** to traverse it.

*   A standard object, like a Python **list**, loads all items into RAM simultaneously.
*   An Iterator works smartly: it loads only one item into memory at a time, performs the required operation (e.g., multiplication or printing), removes it, and then loads the next item.
*   This method is highly **memory efficient**. For example, when generating and operating on numbers from 1 to 100,000, using a Python `range` object (which is an Iterable that generates an Iterator) consumes significantly less memory compared to creating a full list of those numbers.

This efficiency makes the Iterator concept crucial, especially when working with vast amounts of data.

### 3. Distinguishing Iterables and Iterators

It is essential to understand the relationship between these two object types.

#### A. Identifying an Iterable

An object is an Iterable if it supports iteration. You can confirm this in two ways:

1.  **Loop Test:** Try running a `for` loop over the object. If it runs, it is an Iterable.
2.  **`dir()` Function:** Check the object using the `dir()` function. If the magic method **`__iter__`** (the `iter` method) is present, the object is an Iterable.

#### B. Identifying an Iterator

An object is an Iterator only if it has the necessary internal mechanism for state tracking.

*   **`dir()` Function:** Check the object using the `dir()` function. If **both** the magic methods **`__iter__`** (iter) AND **`__next__`** (next) are present, the object is an Iterator.

#### C. Relationship Rules

1.  **An Iterable generates an Iterator**. If you pass an Iterable (e.g., a list, `L`) to the built-in Python function `iter(L)`, it returns an Iterator object (e.g., a `list_iterator`).
2.  **Not all Iterables are Iterators**. A list is an Iterable, but it is not an Iterator because it loads all data into memory at once.
3.  **Every Iterator is also an Iterable**. This is because an Iterator object, by definition, must contain the `__iter__` method.

#### D. The Confusing Point (`iter()` on an Iterator)

When the `iter()` function is called on an Iterable, it generates a new Iterator object. However, when the `iter()` function is called on an existing Iterator object, **it simply returns the Iterator object itself**.

### 4. How the Python `for` Loop Works Internally

The structure of the Python `for` loop, which abstracts away complex indexing and termination logic, is achieved by relying on Iterators.

When a `for` loop executes on an Iterable (e.g., a list):

1.  **Fetching the Iterator:** Python first calls the `iter()` function on the Iterable to fetch the associated Iterator object.
2.  **Repeatedly Calling `next()`:** The loop then repeatedly calls the `next()` function on this Iterator.
    *   The `next()` function tracks the Iterator’s internal state and returns the subsequent item in the sequence.
3.  **Termination:** When all items have been retrieved, the Iterator runs out of data. The next call to `next()` raises a **`StopIteration` exception**. This exception is caught internally by the `for` loop, causing it to terminate gracefully.

This internal logic can be demonstrated by creating a custom `for` loop function using a `while True` loop combined with `try-except` blocks to handle the `StopIteration` exception.

### 5. Building Custom Iterables and Iterators (The Range Example)

To enable looping functionality over a custom object, one must implement the required Iterable and Iterator protocols. This usually involves creating two linked classes, as demonstrated by building a custom `MeraRange` object that behaves like the built-in Python `range` function.

*   **The Iterable Class (`MeraRange`):**
    *   Requires an `__init__` method to store starting and ending values.
    *   Must include the **`__iter__`** method. This method returns an object of the companion Iterator class (`MeraRangeIterator`).

*   **The Iterator Class (`MeraRangeIterator`):**
    *   Must include both the **`__iter__`** and **`__next__`** methods.
    *   The `__iter__` method simply returns `self` (the Iterator object itself).
    *   The **`__next__`** method contains the main logic:
        *   It checks if the current position (start value) has passed the end value.
        *   If the condition fails (start is greater than or equal to end), it **raises the `StopIteration` exception**.
        *   If the condition is met, it stores the current value, increments the start value for the next call, and returns the current value.

### 6. Application Areas

The primary benefit of Iterators—processing data streams without full memory loading—is critical in fields dealing with large datasets.

*   **Data Science and Machine Learning:** Iterators are crucial when working with datasets (like streams of image data) that are too large to fit entirely into the system's RAM (e.g., 8 GB).
*   **Handling Big Data:** Iterators can be used to create efficient data pipelines that fetch data one item at a time (e.g., reading an image file), convert it (e.g., to a NumPy array), perform operations, and then discard it before loading the next item.
*   Libraries like Keras use Iterator-based concepts, such as **Image Data Generators**, for efficient handling of large image repositories.

```python
# creating my own for loop

def my_for_loop(l):
    iterator=iter(l)
    while True:
        try:
            print(next(iterator))
        except   StopIteration:
             return
```

```python
my_for_loop(l)
```

```python
# range function returns all element between start and end
# our own range class

# range is an iterable -> it must have __iter__
class my_range:
    def __init__(self,start,end):
        self.start=start
        self.end=end

    def __iter__(self):
        return my_range_iterator(self)

# my_range_iterator  is an iterator -> it must have __next__ and __iter__
class my_range_iterator:
    def __init__(self,iterable):
        self.iterable=iterable

    def __iter__(self):
        return self

    def __next__(self):
        if self.iterable.start>=self.iterable.end:
            raise StopIteration
        else:
            self.iterable.start+=1
            return self.iterable.start-1
```

```python
for i in my_range(1,10):
    print(i)
```
