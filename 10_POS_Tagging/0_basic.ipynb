POS tagging means we tag each word in our sentence as pos like for eg-

I am a boy => I->pronoun etc

benefits of Pos tagging -

1. Improves Understanding of Sentence Structure

2. Essential for Many NLP Tasks

3. Enables Word Sense Disambiguation (WSD)

4. Facilitates Dependency and Constituency Parsing etc

# Implimentation using spacy

```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(u"I will google about facebook")
```

```python
doc.text
```

```python
doc[-1],doc[0]  # words can be accesesd as elements of list
```

```python
doc[0],doc[0].pos_    # it tell pos of words
```

```python
doc[0],doc[0].tag_     # it tell more exact pos
```

```python
spacy.explain('PRP')    # for explanation of tag_
```

```python
for word in doc:
    print(word.text,"------>", word.pos_,word.tag_,spacy.explain(word.tag_))
```

```python
from spacy import displacy
```

```python
displacy.render(doc,style='dep',jupyter=True)
```
