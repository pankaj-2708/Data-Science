# 🚀 APIs & FastAPI Explained

## 📌 How Do APIs Handle Requests?

APIs typically involve **three key components**:

1. **Web Server**  
   The interface through which the user (or client) sends an HTTP request (e.g., via a browser, app, or Postman).

2. **Gateway Layer (ASGI / WSGI)**  
   Converts raw HTTP requests into a format the Python application can process.

   - **WSGI (Web Server Gateway Interface)** → Traditional, **synchronous**, used by Flask, Django.
   - **ASGI (Asynchronous Server Gateway Interface)** → Modern, **asynchronous**, used by FastAPI, Starlette.

3. **API Logic**  
   Your Python code — receives the request, processes it, interacts with databases/models, and returns a response.

## ⚡ Why Is FastAPI So Fast?

### 1. ✅ Native Async Support

- FastAPI fully supports Python's `async` / `await`.
- This allows **non-blocking I/O**, meaning it can **handle thousands of requests concurrently**.
- In contrast, older frameworks like Flask or Django use **WSGI**, which processes one request at a time (unless extra setups like Gunicorn workers or asyncio hacks are added).

```python
@app.get("/data")
async def get_data():
    data = await fetch_from_db()
    return data
```

---

### 2. 🚀 Powered by ASGI & Starlette

- FastAPI uses **Starlette**, a high-performance ASGI framework.
- **ASGI** (Asynchronous Server Gateway Interface) enables **true async processing**, unlike **WSGI**, which is synchronous and blocking.

> This switch from WSGI → ASGI is one of the core reasons for FastAPI’s performance boost.

---

### 3. ⚡ Pydantic for Fast Validation

- FastAPI uses **Pydantic** for request parsing and validation, based on Python type hints.
- It’s **Cython-optimized**, meaning it parses and validates data **blazingly fast**.
- Less code, fewer bugs, more speed!

```python
from pydantic import BaseModel

class Item(BaseModel):
    name: str
    price: float
```

---

### 4. ⏱️ Fast to Code, Fast to Deploy

- FastAPI **auto-generates documentation** (Swagger UI & Redoc).
- Built-in **dependency injection**, request parsing, and response modeling.
- Eliminates boilerplate code — making development quicker and more maintainable.
