## Lecture Notes: Word2Vec (NLP - Word Embeddings)

### 1. Introduction to Word Embeddings

**Word Embeddings** are techniques used for the representation of words or text. The primary purpose is to convert words into real-valued vectors (collections of numbers) so that machines can process them and apply algorithms.

*   **Goal:** The representation must capture the meaning of the word such that words close to each other in vector space are expected to be similar in meaning.
*   **Categories of Word Embedding Techniques**:
    1.  **Frequency-Based:** These techniques determine vector numbers based on frequency counts.
        *   *Examples:* Bag-of-Words (BoW), TF-IDF, and GloVe (Global Vectors for Word Representation, based on matrix factorization).
    2.  **Prediction-Based:** The conversion of words to vectors is based on the prediction of an underlying algorithm, typically Deep Learning/Neural Networks.
        *   *Example:* **Word2Vec**.

### 2. Word2Vec Overview

Word2Vec is a highly regarded word embedding technique.

*   **Origin:** It was developed by engineers at Google in 2013.
*   **Necessity:** Word2Vec is required because words cannot be processed directly, nor can algorithms be applied directly to them; numerical representations are needed.
*   **Advantages over Frequency-Based Techniques (BoW, TF-IDF)**:

| Feature | Older Techniques (BoW/TF-IDF) | Word2Vec |
| :--- | :--- | :--- |
| **Semantic Meaning** | Cannot capture semantic meaning (e.g., that "happy" and "joy" are similar). | **Can capture semantic meaning** and relationships. |
| **Vector Dimensionality** | High-dimensional (e.g., 10,000 dimensions). | **Low-dimensional** (typically 100 to 300 dimensions). |
| **Vector Structure** | **Sparse** (many zero values), leading to potential overfitting. | **Dense** (non-zero values), preventing overfitting problems. |
| **Computation** | Slower due to high dimensionality. | **Faster** due to low dimensionality.

### 3. Word2Vec Intuition and Working Logic

#### A. Vectors as Features

Internally, Word2Vec creates **features** for each word.

*   A vector (e.g., 300 numbers) is a collection of values representing abstract features (like F1, F2, F3, etc.).
*   While we can imagine features like 'Gender' or 'Wealth' for a small vocabulary, for real-world vocabularies (e.g., 3 million words), **hand-crafting these features is impossible**.
*   Neural Networks are used to automatically create these features. A downside is that the specific semantic meaning of these automatically generated features (e.g., what F1 represents) is often unknown.

#### B. The Core Principle: Context

The fundamental logic used by Word2Vec to create these features is based on context:

*   **"Two words sharing similar contexts also share similar meaning and consequently a similar vector representation from the model."**.
*   If "football" and "hockey" are frequently used in similar surrounding words (context), Word2Vec will learn they are semantically close (e.g., similar sports).

#### C. Vector Arithmetic and Similarity

Since word embeddings are vectors, standard vector operations can be applied.

*   **Similarity:** The similarity between two words can be calculated using their vectors, typically via **cosine similarity**.
*   **Vector Arithmetic:** Vectors can be added and subtracted to reveal linguistic analogies.
    *   *Example:* Calculating King - Man + Woman should result in a vector representation that is closest to Queen.
    *   *Example:* INR (India's currency) - India + England results in vectors close to GBP (Great British Pound).

### 4. Word2Vec Architectures

Word2Vec relies on a **dummy problem** to train a neural network; the resulting word vectors are a **byproduct** of solving this problem. There are two main architectures:

#### 4.1. Continuous Bag-of-Words (CBOW)

*   **Dummy Problem:** **Given the context words, predict the target word** (the middle word).
*   **Training Data Creation:** A fixed-size window (e.g., 3 words: C1, T, C2) is slid across the corpus. The input consists of the context words (C1, C2), and the expected output is the target word (T).
*   **Neural Network Structure:**
    *   Input: Typically one-hot vectors representing the context words.
    *   Hidden Layer: The number of nodes in the hidden layer determines the desired **dimension** of the output vector (e.g., 3 nodes for a 3-dimensional vector).
    *   Output Layer: Nodes equal to the size of the total vocabulary, often using a Softmax layer to predict the probability of the target word.
*   **Vector Extraction:** After training the network across multiple epochs to minimize the loss, the **weights** connecting the input layer to the hidden layer become the final word vectors (embeddings).

#### 4.2. Skip-gram

*   **Dummy Problem:** **Given the target word, predict the context words**.
*   **Structure:** Skip-gram reverses the architecture of CBOW. The target word is the input, and the context words (C1, C2) are the desired outputs.
*   **Vector Extraction:** Similar to CBOW, the embeddings are extracted from the weights of the trained neural network.

#### 4.3. Architecture Selection Guidance

*   **Small Data:** Use **CBOW** (it is faster and performs well on smaller datasets).
*   **Large Data:** Use **Skip-gram** (it tends to yield better results on larger datasets).

### 5. Training and Improving Word2Vec Quality

#### A. Training a Custom Model (Example: Gensim Library)

The Gensim library is often used to train Word2Vec models.

1.  **Data Preparation:** The corpus (text data) must be processed, typically into a list of lists, where the outer list contains sentences, and the inner lists contain tokenized words. Stop word removal is recommended for cleaner results.
2.  **Configuration:** Define model parameters:
    *   `window`: How many words on either side of the target word are considered context (e.g., 10).
    *   `vector_size`: The desired dimension of the output vectors (e.g., 100 or 300).
    *   `min_count`: The minimum frequency a word must have to be included in the vocabulary.
3.  **Training:** The `model.train()` function is called, specifying the corpus, the total number of examples (sentences), and the number of `epochs` (how many times the model iterates over the data).

#### B. Quality Improvement Factors

The quality of the Word2Vec embeddings can be improved by:

1.  **Increasing the training data.**
2.  **Increasing the dimension of the vectors** (i.e., increasing the number of nodes in the hidden layer).
3.  **Increasing the window size** (Note: this often increases training time and complexity).

#### C. Visualizing Embeddings

Since vectors are typically high-dimensional (e.g., 100D), **dimensionality reduction techniques** must be applied to plot them in 2D or 3D space.

*   Common Techniques: PCA (Principal Component Analysis) or t-SNE.
