There are two type of vectorization techniques 

1- Bag of words 

    in this we create a vocoubalary (set of all unique words) and represent each word of vocabulary as an feature and write their count on the column values.
    implemented using countvectorizer in sckit learn


2- Tf-idf

    in this we create a vocoubalary (set of all unique words) and represent each word of vocabulary as an feature and write their tf-idf value on the column values.

    our end goal is to assign some weight to terms so that it can be identified that how much that term is important for a particular row
    
    TF-IDF value

    tf- term frequency (frequency of that term in the given row) = count of that term/total no of terms

    idf- Inverse Document Frequency (frequency of that term in all rows) = log(total no of rows/total no of rows that have that term in them) + 1

        we take log to  scale down that frequency and add 1 so that it won't ignore those terms which occur in all rows
